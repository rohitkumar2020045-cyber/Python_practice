```python
#7. Write a program to find the factorial of a number using a while loop.

n = int(input("Enter a num to find the factorial: "))
fact = 1
if n < 0:
    print("Factorial does not exist for negative numbers")
elif n ==0:
    print("The factorial of 0 is 1")
else:
    while n >0:
        fact *= n
        n -= 1
    print("The factorial is", fact)
    
```

```python
#10. Write a program to print the Fibonacci series up to n terms using a while loop.

n = int(input("Enter the number of terms for fibonacci series: "))
a = 0
b = 1
count = 0
if n <= 0:
    print("Enter a positive integer")
elif n == 1:
    print("Fibonacci series upto ", n, ":")
    print(a)
else:
    print("Fibonacci series: ")
    while count < n:
        print(a)
        a = b
        b = (a+b)
        count += 1

```

```python
#15. Write a function count_vowels(word) that returns the number of vowels in a given word.

def count_vowels(word):
    count = 0
    vowels = "aeiouAEIOU"
    for char in word:
        if char in vowels:
            count += 1
    return count

count_vowels("Manish kumar")


    
```
