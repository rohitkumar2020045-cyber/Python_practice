```python
#Singly Linked list

class node:
    def __init__(self,data):
        self.data =data   #store the value
        self.next =None   #store the reference of the next node

class linkedlist:
    def __init__():
        self.head = None   #keeps track of start of the list

node1 = node(10)
node2 = node(20)
node3 = node(30)
node4 = node(40)

node1.next = node2   #node1 point to the node2 address
node2.next = node3
node3.next = node4

#set the head node
head= node1

#temporary node
current = head
while current is not None:

    print(current.data, end=" -> ")
    current = current.next 
print("None")


```

```python
#Doubly Linkedlist  

class node:
    def __init__(self,data):
        self.data =data
        self.prev =node   #store the value
        self.next =None   #store the reference of the next node

class linkedlist:
    def __init__():
        self.head = None   #keeps track of start of the list

node1 = node(10)
node2 = node(20)
node3 = node(30)


node1.next = node2   #node1 point to the node2 address
node2.prev = node1
node2.next = node3
node3.prev = node2


#new node insertion between 20 and 30
new_node = node(25)
new_node.next = node3
new_node.prev = node2
node2.next = new_node

#new node insertion between 10 and 20
new_node2 =node(15)
new_node2.next =node2
new_node2.prev =node1
node1.next =new_node2


current = node1
while current is not None:

    print(current.data, end=" -> ")
    current = current.next 
print("None")
```

```python
class node:
    def __init__(self,data):
        self.data = data
        self.prev = node
        self.next = None

node1 = node(22)
node2 = node(33)
node3 = node(44)
node4 = node(55)


node1.next = node2
node2.prev = node1

node2.next = node3
node3.prev = node2

node3.next = node4
node4.prev = node3

#deletion of node3

node3.prev.next = node3.next
node3.next.prev = node3.prev


current = node1

while current is not None:
    print(current.data, end=" -> ")
    current = current.next
print(None)
```
